```python
import statsmodels.api as sm
```

```python
X=dataset['dosage_mg']
y=dataset['recovery_days']
X=sm.add_constant(X)
```

```python
model=sm.OLS(y,X).fit()
```

```python
print(model.summary())
```

```python
r_squared = model.rsquared
print(f"R²: {r_squared:.4f}")
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
intercept = model.params[0]
slope = model.params[1]
print(f"Estimated regression equation: dosage_mg= {intercept:.2f} + {slope:.2f} * recovery_days")
```
